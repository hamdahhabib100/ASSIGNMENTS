```python
df = df.dropna(subset=["YIELD_KG"])
df
```

```python
df["AREA_ACRES"] = df["AREA_ACRES"].fillna(df["AREA_ACRES"].mean())
df
```

```python
df["DISTRICT"] = df["DISTRICT"].fillna("Unknown")
df
```

```python
df["FERTILIZER_USED"] = df["FERTILIZER_USED"].fillna("No")
df
```

```python
df["YIELD_VALUE"] = df["YIELD_VALUE"].fillna(df['YIELD_VALUE'].mean())
df
```

```python
df["HARVEST_DATE"] = pd.to_datetime(df["HARVEST_DATE"], format="%d/%m/%Y", errors="coerce")
df
```

```python
df["FARM_ID"] = (df["FARM_ID"].astype(str).str.replace("F", "").astype(int))
df
```

```python
yield_by_crop = df.groupby('CROP_TYPE').agg({'YIELD_KG': 'sum'})
yield_by_crop
```

```python
df
```

```python
yield_by_crop.plot(kind="bar",title="Total Yield by Crop Type (KG)",
    xlabel="Crop Type",
    ylabel="Yield (KG)",
    legend=False,
    rot=0,
    width=0.5,
    alpha=0.75);
```

```python
yield_by_date = df.pivot_table(
    values="YIELD_KG",
    index="HARVEST_DATE",
    aggfunc="sum",
)
yield_by_date
```

```python
plt.plot(yield_by_date.index, yield_by_date["YIELD_KG"])
plt.title("Yield Trend Over Time (KG)")
plt.ylabel("Yield (KG)")
plt.savefig("crop_yield_trend.png")
plt.show()
```
