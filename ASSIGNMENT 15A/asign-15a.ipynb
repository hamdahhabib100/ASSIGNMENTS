# Q1: AVG FARE BY SERVICE TYPE

```python
df.groupby("service_type")[["fare"]].agg(["mean"]).round(2)
```

# Q4: COMPARISON OF TRIP RATINGS BY GENDER

```python
df.groupby(["gender"])[["trip_rating"]].agg(["sum","mean","median","min","max","count"]).round(2)
```

# Q5: BAR CHART (SERVICE TYPE VS. AVG FARE)

```python
df.groupby("service_type")[["fare"]].agg(["mean"]).round(2).plot(
    kind="bar",
    title="Average Fare by Service Type",
    xlabel="Service Type",
    ylabel="Average Fare",
    legend=False, rot=0 ,
    width=0.5, alpha=0.6
);
```
