```python
import pandas as pd

# 1. loading file
file_path = r"C:\Users\HP\Downloads\teaching_prices_cleaned_audit.csv"
df = pd.read_csv(file_path)

# 2. Mean and Median :-
print("--- Mean ---")
print(df.mean(numeric_only=True))

print("\n--- Median ---")
print(df.median(numeric_only=True))

# 3. First-to-Last Percentage Change 
numeric_df = df.select_dtypes(include='number')
pct_change = ((numeric_df.iloc[-1] - numeric_df.iloc[0]) / numeric_df.iloc[0]) * 100

print("\n--- First-to-Last Percentage Change ---")
print(pct_change)
```
